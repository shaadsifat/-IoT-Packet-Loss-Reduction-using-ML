# Combine UNSW-NB15 Raw CSV Parts

The four `UNSW-NB15_*.csv` files under `raw_data/` contain the raw records with **no header row**. The column names live separately in `raw_data/NUSW-NB15_features.csv`. This notebook:

1. Loads the feature list and builds the column names.
2. Reads and concatenates the 4 raw parts.
3. Applies the column headers.
4. Saves the combined result as `unsw_combined_dataset.csv` in the project folder (outside `raw_data/`).

```python
import pandas as pd
from pathlib import Path

RAW_DATA_DIR = Path("raw_data")
OUTPUT_PATH = Path("unsw_combined_dataset.csv")

raw_files = [
    RAW_DATA_DIR / "UNSW-NB15_1.csv",
    RAW_DATA_DIR / "UNSW-NB15_2.csv",
    RAW_DATA_DIR / "UNSW-NB15_3.csv",
    RAW_DATA_DIR / "UNSW-NB15_4.csv",
]
features_path = RAW_DATA_DIR / "NUSW-NB15_features.csv"
```

## Column names

Defined directly from `NUSW-NB15_features.csv` (49 columns, in order).

```python
column_names = [
    "srcip", "sport", "dstip", "dsport", "proto", "state", "dur",
    "sbytes", "dbytes", "sttl", "dttl", "sloss", "dloss", "service",
    "Sload", "Dload", "Spkts", "Dpkts", "swin", "dwin", "stcpb", "dtcpb",
    "smeansz", "dmeansz", "trans_depth", "res_bdy_len", "Sjit", "Djit",
    "Stime", "Ltime", "Sintpkt", "Dintpkt", "tcprtt", "synack", "ackdat",
    "is_sm_ips_ports", "ct_state_ttl", "ct_flw_http_mthd", "is_ftp_login",
    "ct_ftp_cmd", "ct_srv_src", "ct_srv_dst", "ct_dst_ltm", "ct_src_ltm",
    "ct_src_dport_ltm", "ct_dst_sport_ltm", "ct_dst_src_ltm", "attack_cat",
    "Label",
]

print(len(column_names), "columns defined")
```

Sanity check against the features file (mainly to confirm the count of 49 columns).

```python
features_df = pd.read_csv(features_path, encoding="cp1252")
print(features_df.shape)
features_df.head()
```

## Read and combine the 4 raw parts

The raw files have no header, so `header=None` is used and the column names are assigned afterwards. `encoding="utf-8-sig"` strips the BOM present at the start of the first file.

```python
dfs = []
for f in raw_files:
    df_part = pd.read_csv(
        f,
        header=None,
        names=column_names,
        low_memory=False,
        encoding="utf-8-sig",
    )
    print(f.name, df_part.shape)
    dfs.append(df_part)

combined_df = pd.concat(dfs, ignore_index=True)
combined_df.shape
```

## Save combined dataset

```python
combined_df.to_csv(OUTPUT_PATH, index=False)
print(f"Saved combined dataset to: {OUTPUT_PATH.resolve()}")
```
